# What drives underpricing in Indian IPOs?

**Dataset:** 509 Indian IPOs from January 2010 to August 2025, with 19 columns: offer details, subscription levels, market conditions and the underpricing result. 
Source: [https://www.kaggle.com/datasets/ishans474/indian-ipo-underpricing-2010-2025].

**Key columns:**
- Issue_Size(crores): size of the issue
- QIB, HNI, RII, Total: subscription multiples for institutional investors, high-net-worth investors, retail investors and overall
- NIFTY_20D_Return, NIFTY_60D_Return: Nifty index returns over 20 and 60 days after the IPO date (market conditions, not the IPO's own return; verify in the dataset description)
- India_VIX: market fear index at the time of the IPO
- Underpricing: percentage gain at listing compared to the offer price (verify in the dataset description)
- Underpricing_Class: Low / Medium / High category of underpricing

**Main question:** Are heavily subscribed IPOs more underpriced, and do market conditions (VIX) make a difference?

**Tools:** Python, pandas.

```python
#to know how many are launched everyyear
df['Date'] = pd.to_datetime(df['Date'])
ipo_counts = df['Date'].dt.year.value_counts().sort_index()
print(ipo_counts)

#IPO activity varied a lot: 72 in 2010, a low of 4 in 2013, and a high of 75 in 2024. Activity dropped sharply from 2012 to 2014 and recovered from 2021 onwards. 
# 2025 is a partial year (data ends in August).
```

```python
# to know how many are energies IPOs
energy_ipos = df[df['IPO_Name'].str.contains('Energ', case=False, na=False)]
print(energy_ipos[['IPO_Name', 'Issue_Size(crores)']])

# 9 IPOs have "Energ" in their name, such as JSW Energy and NTPC Green Energy.
#  This is only a rough proxy because the dataset has no sector column: it misses energy companies without the word in their name and includes companies like Ather Energy (electric scooters).
```

```python
df["Date"] = pd.to_datetime(df["Date"])
df["Year"] = df["Date"].dt.year
```

```python
# to find Average underpricing by year (groupby) .
df.groupby("Year")["Underpricing"].agg(["count", "mean", "median"]).round(2)

#Median underpricing by year: 
#- Highest in 2020 (49%), but only 13 IPOs, so this is a small sample.
#- Negative in 2018 (-0.14%) and exactly 0% in 2015, meaning many IPOs listed flat or lower.
#- In several years the mean is much higher than the median (2017: 20.7% vs 3.1%; 2021: 24.5% vs 5.7%). A few very large listings pull the mean up, so the median describes the typical IPO better.
#- 2013 (4 IPOs) and 2014 (6 IPOs) are too small to draw conclusions from.
```

```python
df[["Underpricing", "Total", "QIB", "India_VIX", "Issue_Size(crores)"]].corr()

#Underpricing is strongly positively correlated with overall subscription (0.72) and QIB subscription (0.60).
#  It has a weak negative link with India VIX (-0.17) and issue size (-0.10). Total and QIB are themselves strongly correlated (0.81), because institutional demand is part of total demand, so they should not be treated as independent evidence. 
```

```python
df["Sub_Group"] = pd.qcut(df["Total"], q=3, labels=["Low", "Medium", "High"])
```

```python
df.groupby("Sub_Group", observed=True)["Total"].agg(["count", "min", "max"])

#IPOs were split into three equal groups by total subscription: Low (up to 2.97 times), Medium (2.98 to 30.57 times) and High (30.63 to 326 times).
```

```python
# to find if high-subscription IPOs have higher underpricing?
df.groupby("Sub_Group", observed=True)["Underpricing"].agg(["count", "mean", "median"]).round(2)

#Underpricing rises steadily with subscription. Median underpricing is 0% in the Low group, 4.3% in the Medium group and 31.2% in the High group (means: -0.8%, 7.4%, 41.5%).
#  Heavily subscribed IPOs are, on average, listed at much higher prices than their offer price. The High group covers a wide range (30x to 326x), so there is variation inside it.
```

```python
# to find if IPOs in high-VIX periods show lower underpricing
df["VIX_Group"] = pd.qcut(df["India_VIX"], q=3, labels=["Low", "Medium", "High"])
df.groupby("VIX_Group", observed=True)["Underpricing"].agg(["count", "mean", "median"]).round(2)

#IPOs were split into three equal groups by India VIX. Median underpricing falls as market fear rises: 10.6% in Low-VIX periods, 6.0% in Medium and 2.9% in High.
#  IPOs in calmer markets tend to be more underpriced. The effect is smaller than the subscription effect.
```

## Conclusions

1. **Subscription level is the strongest signal.** Median underpricing is 0% for IPOs subscribed under 3 times, versus 31% for those subscribed over 30 times.
2. **Market conditions matter, but less.** Median underpricing falls from 10.6% in low-VIX periods to 2.9% in high-VIX periods.
3. **Issue size shows little link** to underpricing (correlation -0.10). The largest IPO, LIC, listed below its offer price.
4. **Average figures are misleading.** A few extreme listings push the mean well above the median in many years, so medians are used throughout.

## Limitations
- These are associations, not proof of cause. Strong demand and high listing gains may both reflect the same market enthusiasm.
- Some years have very few IPOs (2013: 4, 2014: 6), and 2025 ends in August.
- Subscription and VIX are related to each other, which makes their effects hard to fully separate.
- 509 IPOs from a single dataset is a modest sample, and the energy filter is based only on company names.
